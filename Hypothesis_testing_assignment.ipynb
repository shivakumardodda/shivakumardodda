**Hypothesis Testing Process**

**1.Hypothesis testing:**Hypothesis testing is the statistical method which is used by analyst to test on assumption regarding the population data by using sample data.

H0=Null hypothesis.
H1=Alternative hypothesis.

Sample data:The data which is subset of the population data and its a peice of the data which is cost effective and time saving process

**2.Caluclate test statistic:**

Assumption:

H0=Cost model not accurate

H1=Cost model is accurate

find the which is correct.

if t-statistical value>t-critical value (H1 is true)

if t-statistical value<t-critical value(H0 is true)

```python
# Importing Libraries
from scipy import stats
from scipy.stats import norm
import math
```

```python
#Data Provided:
sample_mean = 3050  # in Rs.
theoretical_mean = 4000  # in Rs.
std_dev = 5 * 25  # in Rs.
sample_size = 25
alpha = 0.05
```

```python
#2 Calculate the test statistic
standard_error = std_dev / math.sqrt(sample_size)
test_statistic = (sample_mean - theoretical_mean) / standard_error
test_statistic
```

```python
# Degrees of freedom
dof = sample_size - 1
dof
```

**3.Determine the critical value:**

alpha value=0.05
degress of freedom(df)=n-1=24

```python
from scipy import stats
t_critical=stats.t.ppf(1-0.05,24)
t_critical
```

By comparing the t_stastical value and t_critical value,we get to know about the model is working accurate or not.whether your null hypothesis is to be rejected or not
